```python
import requests
import pandas as pd
from datetime import datetime, timedelta

def fetch_delivery_data(date_str, token):
    url = "https://tms.tdccorp.com.tw/webservice/WsTDC.asmx/GetTask_MasterData"

    headers = {
        "accept": "application/json, text/javascript, */*; q=0.01",
        "content-type": "application/x-www-form-urlencoded; charset=UTF-8",
        "authorization": token,
        "x-requested-with": "XMLHttpRequest",
        "Referer": "https://tms.tdccorp.com.tw/web/MoveManage/DeliveryDynamic.aspx"
    }

    data = {
        "Delivery_Dt": date_str,
        "Dept": "RF",
        "TaskType_id": "",
        "KeyWord": ""
    }

    response = requests.post(url, headers=headers, data=data)
    result_raw = response.json()

    df = pd.DataFrame(result_raw["DATA"])
    df["Delivery_Dt"] = date_str  # 加入日期欄位
    return df
```

```python
# 查詢區間
start_date = datetime.strptime("2024-01-01", "%Y-%m-%d")
end_date = datetime.strptime("2025-03-31", "%Y-%m-%d")

# 抓資料
all_data = []
current = start_date
while current <= end_date:
    date_str = current.strftime("%Y-%m-%d")
    try:
        print(f"📦 擷取 {date_str} 中...")
        df_day = fetch_delivery_data(date_str, auth_token)
        all_data.append(df_day)
    except Exception as e:
        print(f"{date_str} 失敗：{e}")
    current += timedelta(days=1)

# 合併儲存
if all_data:
    df_all = pd.concat(all_data, ignore_index=True)
    cols = ["TaskType_id", "Delivery_Dt", "Dept_Name", "Route_seq", "Car_licence", "Store_id", "Store_Name", "Pl_M_Arrive_dt", "Ac_Arrive_dt", "Im_Arrive_dt"]
    filename = f"{start_date.strftime('%Y-%m-%d')}_{end_date.strftime('%Y-%m-%d')}.csv"
    df_all[cols].to_csv(filename, index=False, encoding="utf-8-sig")
    print(f"已儲存為 {filename}")
else:
    print("⚠️ 沒有成功取得任何資料")
```

```python
# 按照配別名稱分類
import os
import pandas as pd

folder_path = "route"
csv_files = [f for f in os.listdir(folder_path) if f.endswith(".csv")]

# 讀取並合併所有資料
all_data = []
for filename in csv_files:
    file_path = os.path.join(folder_path, filename)
    try:
        df = pd.read_csv(file_path)
        df["source_file"] = filename
        all_data.append(df)
    except Exception as e:
        print(f"❌ 讀取失敗：{filename} - {e}")

# 合併所有 DataFrame
if all_data:
    df_all = pd.concat(all_data, ignore_index=True)

    # 篩選 TaskType_id 為 02、03、04、05、12
    df_filtered = df_all[df_all["TaskType_id"].astype(str).isin(["02", "03", "04", "05", "12"])]

    # 分類儲存
    output_paths = {}
    for task_type, group in df_filtered.groupby("TaskType_id"):
        output_path = f"route/route_TaskType_{task_type}.csv"
        group.to_csv(output_path, index=False, encoding="utf-8-sig")
        output_paths[task_type] = output_path
else:
    output_paths = {}

output_paths
```

```python
# 按照車牌分類
import os
import pandas as pd

# 設定資料夾
folder_path = "route"

# 取得 CSV 檔案列表（使用 os）
csv_files = [f for f in os.listdir(folder_path) if f.endswith(".csv")]

# 合併資料
all_data = []
for file in csv_files:
    file_path = os.path.join(folder_path, file)
    try:
        df = pd.read_csv(file_path)
        df["source_file"] = file
        all_data.append(df)
    except Exception as e:
        print(f"❌ 無法讀取 {file}: {e}")

# 合併為單一 DataFrame
df_all = pd.concat(all_data, ignore_index=True)

# 分車牌輸出
output_paths = {}
for car, group in df_all.groupby("Car_licence"):
    if pd.isna(car) or str(car).strip() == "":
        continue  # 跳過空車牌
    safe_car = str(car).replace("/", "_").replace("\\", "_")
    filename = f"route/route_Car_{safe_car}.csv"
    group.to_csv(filename, index=False, encoding="utf-8-sig")
    output_paths[car] = filename

output_paths
```
